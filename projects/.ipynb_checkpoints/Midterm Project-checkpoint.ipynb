```python
import numpy as np
import matplotlib.pyplot as plt
```

Setup

```python
def tridiagonal_matrix(x, y):
    
    n = len(x)
    h = np.diff(x)  # Step sizes between the data points
    
    
    # Initializing the matrix coefficients
    
    A = np.zeros(n - 2)  # Subdiagonal
    B = np.zeros(n - 2)  # Main diagonal
    C = np.zeros(n - 2)  # Superdiagonal
    D = np.zeros(n - 2)  # Right-hand side vector
    
    
    # Populate the tridiagonal matrix coefficients
    
    for i in range(1, n - 1):
        
        A[i - 1] = h[i-1]             # Subdiagonal
        B[i - 1] = 2 * (h[i-1] + h[i]) # Main diagonal
        C[i - 1] = h[i] # Superdiagonal
        D[i - 1] = 6 * ((y[i+1]-y[i])/h[i] - (y[i]-y[i-1])/h[i-1]) # Right-hand side vector
        
    return A, B, C, D



def thomas_algorithm(a, b, c, d):
    
    
    # Going to be solving the tridiagonal system using the Thomas algorithm.
    #a: sub-diagonal elements (length n-1)
    #b: main diagonal elements (length n)
    #c: super-diagonal elements (length n-1)
    #d: right-hand side (length n)
    #Is going to Return: solution vector x of length n
    
    n = len(b)
    
    
    # Forward elimination
    
    for i in range(1, n):
        
        w = a[i-1] / b[i-1]
        b[i] = b[i] - w * c[i-1]
        d[i] = d[i] - w * d[i-1]
        
        
    # Back substitution
    
    x = np.zeros(n)
    x[-1] = d[-1] / b[-1]
    
    for i in range(n-2, -1, -1):
        
        x[i] = (d[i] - c[i] * x[i+1]) / b[i]
        
    return x



def cubic_spline(x, y):
    a, b, c, d = tridiagonal_matrix(x, y)
    M_inner = thomas_algorithm(a.copy(), b.copy(), c.copy(), d.copy())    
    M = np.array([0.0, *M_inner, 0.0])

    total_points = int(np.sum((x[1:] - x[:-1]) * 100))
    retx = np.linspace(x[0], x[-1], total_points)
    rety = np.zeros(total_points)

    idx = 0

    for i in range(len(x)-1):
        h = x[i+1] - x[i]
        npts = int(h * 100)
        x_interp = np.linspace(x[i], x[i+1], npts, endpoint=False)

        spline_vals = (
            (M[i]/(6*h)) * (x[i+1]-x_interp)**3 +
            (M[i+1]/(6*h)) * (x_interp-x[i])**3 +
            (y[i]/h - M[i]*h/6) * (x[i+1]-x_interp) +
            (y[i+1]/h - M[i+1]*h/6) * (x_interp-x[i])
        )

        rety[idx:idx+npts] = spline_vals
        idx += npts

    return retx, rety
```

Part A - Cubic Spline Interpolation

```python
#1.

xint, yint = cubic_spline(x,y)

plt.figure()
plt.scatter(x, y) # Plotting the data points
plt.plot(xint, yint, 'red') # Plotting the line itself
plt.title(f"Cubic Spline Degree {4} Interpolation of Data Set")
plt.grid(True)
plt.show()
```

A natural cubic spline fits a cubic polynomial between each pair of consecutive data points. With 20 data points, there are 19 spline segments, and each segment has four coefficients, giving 76 total unknown coefficients. Instead of solving for all coefficients directly, the spline is written in terms of the second derivatives at the data points, denoted say M_i = S"(x_i). Because a natural spline requires the second derivatives at the first and last data points to be zero, there are 18 unknown second derivatives to solve for.

At each interior data point, the spline must satisfy continuity conditions. The function values must match so the curve passes through each point. The first derivatives must match so the slope does not change abruptly between segments. The second derivatives must also match so the curvature is continuous. These conditions ensure the spline is smooth across the entire domain.

The natural spline boundary conditions specify that the second derivative at the first and last data points is zero. This makes the spline behave linearly at the ends and provides the additional equations needed to solve the system.

```python
#2.

def cubic_spline_eval(x, y, x_eval):
    
    a, b, c, d = tridiagonal_matrix(x, y)
    M_inner = thomas_algorithm(a.copy(), b.copy(), c.copy(), d.copy())
    M = np.array([0.0, *M_inner, 0.0])

    y_eval = np.zeros_like(x_eval, dtype=float)

    for i in range(len(x) - 1):
        mask = (x_eval >= x[i]) & (x_eval <= x[i+1])
        xs = x_eval[mask]
        h = x[i+1] - x[i]

        y_eval[mask] = (
            (M[i]/(6*h)) * (x[i+1]-xs)**3 +
            (M[i+1]/(6*h)) * (xs-x[i])**3 +
            (y[i]/h - M[i]*h/6) * (x[i+1]-xs) +
            (y[i+1]/h - M[i+1]*h/6) * (xs-x[i])
        )

    return y_eval

x_all = np.arange(x[0], x[-1] + 1)

y_all = cubic_spline_eval(x,y,x_all)

plt.figure()
plt.scatter(x, y) # Plotting the randomly selected data points
plt.plot(x_all, y_all, 'red') # Plotting the line itself
plt.title(f"Cubic Spline Interpolation of Data Set")
plt.grid(True)
plt.show()
```

The spline constructed from the given subset of months can be evaluated at every month between the first and last data points by computing the cubic polynomial within each interval. The resulting interpolated curve is visually smooth because cubic splines are continuous and have continuous first and second derivatives. Even though the months are not evenly spaced, the spline still produces a smooth curve because the interval lengths are included in the spline equations.

```python
# The natural cubic spline interpolant is mathematically smooth: it is C^2, meaning the
# function, its first derivative, and its second derivative are all continuous across
# the entire domain.
#
# However, when I evaluate the spline at every quarter between 2010 Q1 and 2023 Q4,
# the resulting curve does not look gently smooth. The GDP growth data themselves are
# quite irregular, with sharp changes and at least one extreme outlier (2020 Q2 at
# -28.1%). Because the spline is forced to pass exactly through every data point, it
# must bend aggressively to hit these values.
#
# As a result, the interpolant, while technically smooth in the mathematical sense,
# appears visually jagged or wiggly in several regions. In that sense, the spline is
# not entirely "visually smooth," and its behavior between points may not always look
# economically reasonable, especially near the extreme outlier.
```
