```python
import numpy as np
import time as tm
```

```python
def jacobi(A, b, x0, tol, max_iterations):
    
    n = len(b)
    x = x0.copy()
    
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
            
            # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        
        x = x_new
    return x, max_iterations
```

```python
def gauss_seidel(A, b, x0, tol, max_iterations):
    
    n = len(b)
    x = x0.copy()
    
    for k in range(max_iterations):
        x_new = x.copy()
        
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
            
            # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        
        x = x_new
    return x, max_iterations
```

```python
A = np.array([[3, -0.1, -0.2],[0.1, 7, -0.3],[0.3, -0.2, 10]])
b = np.array([7.85, -19.3, 71.4])

x0 = np.zeros_like(b)
tol1 = 1e-3 #tolerance 1
tol2 = 1e-6 #tolerance 2
tol3 = 1e-9 #tolerance 3
max_iterations = 100

start_time1 = tm.time()
solution, iterations = jacobi(A, b, x0, tol1, max_iterations) #Jacobi iterations
end_time1 = tm.time()

print(f"Jacobi Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {end_time1 - start_time1} seconds")



start_time2 = tm.time()
solution, iterations = gauss_seidel(A, b, x0, tol1, max_iterations) #Gauss-seidel iterations
end_time2 = tm.time()

print(f"\nGauss-Seidel Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {end_time2 - start_time2} seconds")
```
