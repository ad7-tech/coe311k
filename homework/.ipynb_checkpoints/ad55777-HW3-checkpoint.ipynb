```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
def sine(x, terms):
    
    sinappf = 0

    for n in range (terms):

        sinapp = ((-1)**n * x**(2*n+1))/math.factorial((2*n+1))
        sinappf += sinapp
        
    return sinappf
```

```python
xs = np.arange(0,10.1,0.1)

exact = np.sin(xs)
app1 = np.array([sine(x,1) for x in xs])
app2 = np.array([sine(x,2) for x in xs])
app3 = np.array([sine(x,3) for x in xs])
app4 = np.array([sine(x,4) for x in xs])

err1 = np.abs(exact - app1)
err2 = np.abs(exact - app2)
err3 = np.abs(exact - app3)
err4 = np.abs(exact - app4)
```

```python
plt.figure(figsize=(10,6))
plt.plot(xs, err1, label="1 term")
plt.plot(xs, err2, label="2 terms")
plt.plot(xs, err3, label="3 terms")
plt.plot(xs, err4, label="4 terms")

plt.xlabel("x")
plt.ylabel("Absolute Error")
plt.title("Truncation Error of Taylor Series Approximation for sin(x)")
plt.legend()
plt.show()
```
