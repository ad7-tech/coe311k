```python
import random
import numpy as np
    
def montecarlo(s):
    
    counter = 0

    for i in range(0,s):
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        
        if (x*x + y*y <1):
            counter+=1
        
    return (4*counter/s)
```
