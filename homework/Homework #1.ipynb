```python
import random
import numpy as np
import matplot.lib as plt
    
def montecarlo(s):
    
    counter = 0

    for i in range(0,s):
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        
        if (x*x + y*y <1):
            counter+=1
        
    return (4*counter/s)
```

```python
c = np.arange(10,1000,20)

for s in c:
        
    a = montecarlo(int(s)) 
    error = np.abs(a-np.pi)

plt.plot(error)
```
