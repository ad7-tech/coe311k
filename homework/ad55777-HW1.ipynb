```python
import random
import numpy as np
import matplotlib.pyplot as plt
    
def montecarlo(s):
    
    counter = 0

    for i in range(0,s):
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        
        if (x*x + y*y <1):
            counter+=1
        
    return (4*counter/s)
```

```python
c = np.arange(10,10000,20)
errls=[]

for s in c:
        
    a = montecarlo(s) 
    error = np.abs(a-np.pi)
    errls.append(error)

plt.plot(c, errls)
plt.xlabel('Samples')
plt.ylabel('Error')
plt.title('Samples vs. Error')
plt.show()
```
