```python
import numpy as np
import matplotlib.pyplot as plt
import time as tm
import random
```

```python
def naive_matrix_multiplication(A, B): # going to be used to go against A@B and compare
    n = len(A)
    C = [[0] * n for _ in range(n)]
    
    for i in range(n):
        for j in range(n):
            for k in range(n):
            
                C[i][j] += A[i][k] * B[k][j]
    return 
```

```python
n = 3 # Inference: That the optimized time will be < than the naive time
naivetime = np.zeros(8)
opttime = np.zeros(8)
size = np.zeros(8)

for i in range (8): # For loop for iterating the size of the matrices, creating arrays for iterated times and sizes, and comparing the runtimes of optimized and naive matrix multiplication.

    x = n*2**i
    
    A = np.random.uniform(1.0, 10, size=(x, x))
    B = np.random.uniform(1.0, 10, size=(x, x))
    
    start_time1 = tm.time()
    C1 = naive_matrix_multiplication(A, B)
    end_time1 = tm.time()
    runtime1 = end_time1 - start_time1 #getting the difference in start and stop times for naive matrix multiplication
    naivetime[i] = runtime1
    
    size[i] = x


    start_time2 = tm.time()
    C2 = A@B     
    end_time2 = tm.time()
    runtime2 = end_time2 - start_time2 #getting the difference in start and stop times for optimized matrix multiplication
    opttime[i] = runtime2

    
    print("\n",x , 'x', x)
    print('Naive Function time:', runtime1 ,'seconds') # printing runtime for naive function
    print('Optimized Function time:', runtime2 ,'seconds') # printing runtime for optimized function
    
```

```python
plt.plot(size, naivetime) #plotting size on the x axis vs the naive matrix multiplication time on the y axis

plt.title('Naive Growth Rate')
plt.xlabel('Matrix Number')
plt.ylabel('Runtime (seconds)')
plt.show()
```

```python
plt.plot(size, opttime) #plotting size on the x axis vs the optimized matrix multiplication time on the y axis

plt.title('Optimized Growth Rate')
plt.xlabel('Matrix Number')
plt.ylabel('Runtime (seconds)')
plt.show()
```

As shown above, the graphs of the Naive and Optimized matrix multiplcation's times growth rates have very similar shapes, but the optimized matrix multiplication is shown to have a much, much smaller time to execute compared to that of the naive matrix multiplication time for every iteration. This supports my initial inference that optimized will be quicker than th enaive version.
