```python
# Challenge D

def diagonal(A):
    
    a1 = 0
    a = 0
    
    for i in range (len(A)): # fidning the diagonal value for each row
        a1 = A[i][i]
        a2 = 0
        
        for j in range (len(A)): #Summing the all values in the row besides the diagonal
            a2 += A[i][j]
            a -= a1

            if (a1>a): #Checking if diagonal is > the sum of the row - diagonal value
                return A, True
            
    return A, False
```
