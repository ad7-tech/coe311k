```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

Part 1:

```python
# Data points
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T

print ("X: ", X)
print ("\nY: ", Y)
print ("\nA: ", A)
```

```python
# Solve using least squares
m1, c1 = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"\nSlope: {m1}, Intercept: {c1}")
```

```python
#Plot each data point 
#Plot the line y = 0.95*x + 0.59
```

```python
plt.figure()
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.scatter(X,Y) #Plots points

x2 = np.linspace(min(X),(max(X)),100) # for line x values
y2 = 0.95*x2 + 0.59 # line

plt.plot(x2,y2) #Plots the best fit line
```

Part 2:

```python
# Data points
X1 = np.random.rand(5) *10
Y1 = np.random.rand(5) *10

# Stack the design matrix with a column of ones (for the intercept)
A1 = np.vstack([X1, np.ones(len(X1))]).T

print ("X: ", X1)
print ("\nY: ", Y1)
print ("\nA: ", A1)
```

```python
# Solve using least squares
m2, c2 = np.linalg.lstsq(A1, Y1, rcond=None)[0]

print(f"\nSlope: {m2}, Intercept: {c2}")
```

```python
plt.figure()
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.scatter(X1,Y1) #Plots points

x3 = np.linspace(min(X1),(max(X1)),100) # for line x values
y3 = m2*x3 + c2 # line

plt.plot(x3,y3) #Plots the best fit line
```

Part 3:

```python
m = 5
b = 2
x0 = np.arange (0,1.01,0.01) #makes points
error = np.random.uniform(-1,1,len(x0)) # +/- error

y0 = m*x0+b+error

# Stack the design matrix with a column of ones (for the intercept)
A0 = np.vstack([x0, np.ones(len(x0))]).T

print ("X: ", x0)
print ("A: ", A0)
```

```python
# Solve using least squares
m0, c0 = np.linalg.lstsq(A0, y0, rcond=None)[0]


print(f"\nSlope: {m0}, Intercept: {c0}")
```

```python
plt.figure()
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")

plt.scatter(x0,y0) #Plots points

xline = np.linspace(min(x0),(max(x0)),101) # for line x values

yline = m0*x0 + c0 # best fit ine
yline2 = m*x0 + b #actual


plt.plot(xline,yline, 'blue', label = 'Best Fit') #Plots the best fit line
plt.plot(xline,yline2, 'red', label = "Original") #Plots the best our line
plt.legend()
```

the best fit was very similar to our actual original line.
