```python
import numpy as np
import matplotlib.pyplot as plt
```

Part 1:

```python
# Data points
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T

print ("X: ", X)
print ("\nY: ", Y)
print ("\nA: ", A)
```

```python
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"\nSlope: {m}, Intercept: {c}")
```

```python
#Plot each data point 
#Plot the line y = 0.95*x + 0.59
```

```python
plt.figure()
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.scatter(X,Y) #Plots points

x2 = np.linspace(min(X),(max(X)),100) # for line x values
y2 = 0.95*x2 + 0.59 # line

plt.plot(x2,y2) #Plots the best fit line
```
