```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,2*np.pi,50)

plt.plot(x, np.tan(x))
plt.show()
```

```python
plt.subplot(2, 2,1)
plt.scatter(x, x**5, 2, 'r')
plt.subplot(2, 2,2)
plt.plot(x, np.sin(x)**2, 'g')
plt.subplot(2, 2,3)
plt.plot(x, np.cos(x)**2, 'y')
plt.subplot(2, 2,4)
plt.plot(x, np.sin(x)**2 / np.cos(x)**2, 'b')
plt.show()
```
