```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define the parameters
k = 1 # Constant in the equation (e.g., spring constant)
h_vals = [0.01,0.02,0.03,0.05,0.1,0.2,0.3,0.5] # Step size
x_start = 0 # Start of the interval
x_end = 10 # End of the interval
y_0 = 1.0 # Initial condition for y
v_0 = 0.0 # Initial condition for dy/dx

plt.figure(figsize=(10, 5))

for h in h_vals:
    
    # Create arrays to store x, y, and v values
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))

    # Set initial conditions
    y_values[0] = y_0
    v_values[0] = v_0

    # Euler's Method loop
    for i in range(1, len(x_values)):
    
        # Current values
        x_n = x_values[i - 1]
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]
    
        # Update equations
        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)
        
    plt.plot(x_values, y_values, label=f'Position y(x) of h = {h}')
    plt.plot(x_values, v_values, label=f'Velocity v(x) of h = {h}', linestyle='--')

        
# Plot the results
plt.xlabel('x')
plt.ylabel('y and v')
plt.title('Solution of d²y/dx² = -ky using Euler\'s Method')
plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))
plt.tight_layout()
plt.grid(True)
plt.show()
```

when does it become unstable?

Position y(x) and velocity v(x) really to start become noticeably unstable at a step size h = 0.2.




Is it becoming unstable?

As the step size h increases, the euler approximation of the solution of d²y/dx² = -ky gets worse, increasing the error, until it becomes violenetly unstable and diverges.

Set k = 10, h = .01

When you increase k, the system becomes stiffer, so Euler’s method blows up faster and the amplitude grows more quickly. With k = 10 and h = 0.01, the solution still oscillates, but the amplitude gets larger and larger because the higher k makes Euler’s instability show up sooner.


Thoughts on whether Euler's Method can solve this with a larger h?

For this second‑order differential equation, the Euler method cannot remain stable with a larger h. Even small h only delays the instability and will eventually become unstable. A different method is needed to fix this problem.
