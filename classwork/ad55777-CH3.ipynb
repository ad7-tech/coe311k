```python
import random
import numpy as np
import matplotlib.pyplot as plt
    
def montecarlo(s):
    
    counter = 0

    for i in range(0,s):
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        
        if (x*x + y*y <1):
            counter+=1
        
    return (4*counter/s)
```

```python
x = 100
i=0

err1 = []
err2 = []
c = np.linspace(0,100,100)
   
for i in range (0,100):
    
    a = montecarlo(1000) 
    error1 = np.abs(np.pi-a)
    error2 = np.abs(np.pi-a)/np.pi
    
    err1.append(error1)
    err2.append(error2)

plt.figure()
plt.plot(c, err1, 'r')
plt.xlabel('Samples')
plt.ylabel('Err1')
plt.title('Samples vs. Err1')
plt.figure()
plt.plot(c, err2, 'b')
plt.xlabel('Samples')
plt.ylabel('Err2')
plt.title('Samples vs. Err2')
plt.show()
    

    
    

    
```
